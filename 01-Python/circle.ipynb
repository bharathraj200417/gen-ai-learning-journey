```python
def circle():
    circle=((1,5),(2,2),(2,8),(5,1),
            (5,9),(8,2),(8,8),(9,5))
    print("=====CIRCLE=====")
    for i in range(1,10):
        for j in range(1,10):
            if(i,j) in circle:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
circle()
```
