```python
#lAT - NUM - ROW
for i in range(1, 6):
    for j in range(5, i, -1):
        print(" ", end=" ")
    for k in range(0,i):
        print(i, end=" ")
    print()
```

```python
#lAT - NUM - COL
for i in range(1, 6):
    for j in range(5, i, -1):
        print(" ", end=" ")
    for k in range(0,i):
        print(k, end=" ")
    print()
```

```python
#lAT - STAR
for i in range(1, 6):
    for j in range(5, i, -1):
        print(" ", end=" ")
    for k in range(0,i):
        print('*', end=" ")
    print()
```

```python
#lAT - UPPER - ROW
for i in range(1, 6):
    for j in range(5, i, -1):
        print(" ", end=" ")
    for k in range(0,i):
        print(chr(i+64), end=" ")
    print()
```

```python
#lAT - UPPER - COL
for i in range(1, 6):
    for j in range(5, i, -1):
        print(" ", end=" ")
    for k in range(1,i):
        print(chr(k+64), end=" ")
    print()
```

```python
#lAT - LOWER - ROW
for i in range(1, 6):
    for j in range(5, i, -1):
        print(" ", end=" ")
    for k in range(0,i):
        print(chr(i+64+32), end=" ")
    print()
```

```python
#lAT - LOWER - COL
for i in range(1, 6):
    for j in range(5, i, -1):
        print(" ", end=" ")
    for k in range(1,i):
        print(chr(k+64+32), end=" ")
    print()
```
