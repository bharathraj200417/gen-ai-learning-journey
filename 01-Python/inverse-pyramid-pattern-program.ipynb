```python
# IPRMD - NUM - ROW
print("===== IPRMD NUM ROW =====")
for i in range(6, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(1, i):
        print(i, end=" ")
    print()
```

```python
# IPRMD - NUM - COL
print("\n===== IPRMD NUM COL =====")
for i in range(6, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(1, i):
        print(k, end=" ")
    print()
```

```python
# IPRMD - STAR
print("\n===== IPRMD STAR =====")
for i in range(6, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(i):
        print("*", end=" ")
    print()
```

```python
# IPRMD - UPPER - ROW
print("\n===== IPRMD UPPER ROW =====")
for i in range(6, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(i):
        print(chr(i + 64), end=" ")
    print()
```

```python
# IPRMD - UPPER - COL
print("\n===== IPRMD UPPER COL =====")
for i in range(6, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(1, i):
        print(chr(k + 64), end=" ")
    print()
```

```python
# IPRMD - LOWER - ROW
print("\n===== IPRMD LOWER ROW =====")
for i in range(6, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(i):
        print(chr(i + 96), end=" ")
    print()
```

```python
# IPRMD - LOWER - COL
print("\n===== IPRMD LOWER COL =====")
for i in range(6, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(1, i):
        print(chr(k + 96), end=" ")
    print()
```

```python
# IPRMD - NAME - ROW
print("\n===== IPRMD NAME ROW =====")
name = "bharath"
for i in range(len(name), 0, -1):
    for j in range(len(name), i, -1):
        print(" ", end="")
    for k in range(i):
        print(name[i - 1], end=" ")
    print()
```
