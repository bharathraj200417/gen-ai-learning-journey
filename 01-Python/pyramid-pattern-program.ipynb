```python
#PRT - NUM - ROW
for i in range(1, 6):
    for j in range(5, i, -1):
        print("", end=" ")
    for k in range(0,i):
        print(i, end=" ")
    print()
```

```python
#PRT - NUM - ROW
for i in range(1, 6):
    for j in range(5, i, -1):
        print("", end=" ")
    for k in range(0,i):
        print(i, end=" ")
    print()
```

```python
#PRT - NUM - COL
for i in range(1, 6):
    for j in range(5, i, -1):
        print("", end=" ")
    for k in range(0,i):
        print(k, end=" ")
    print()
```

```python
#PRT - STAR
for i in range(1, 6):
    for j in range(5, i, -1):
        print("", end=" ")
    for k in range(0,i):
        print("*", end=" ")
    print()
```

```python
#PRT - UPPER - ROW
for i in range(1, 6):
    for j in range(5, i, -1):
        print("", end=" ")
    for k in range(0,i):
        print(chr(i+64), end=" ")
    print()
```

```python
#PRT - UPPER - COL
for i in range(1, 6):
    for j in range(5, i, -1):
        print("", end=" ")
    for k in range(1,i):
        print(chr(k+64), end=" ")
    print()
```

```python
#PRT - LOWER - ROW
for i in range(1, 6):
    for j in range(5, i, -1):
        print("", end=" ")
    for k in range(0,i):
        print(chr(i+64+32), end=" ")
    print()
```

```python
#PRT - LOWER - COL
for i in range(1, 6):
    for j in range(5, i, -1):
        print("", end=" ")
    for k in range(1,i):
        print(chr(k+64+32), end=" ")
    print()
```
