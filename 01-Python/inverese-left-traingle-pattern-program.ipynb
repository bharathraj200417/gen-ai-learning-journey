```python
# ILAT - NUM - ROW
print("===== ILAT NUM ROW =====")
for i in range(5, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(i):
        print(i, end="")
    print()
```

```python
# ILAT - NUM - COL
print("\n===== ILAT NUM COL =====")
for i in range(6, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(1, i):
        print(k, end="")
    print()
```

```python
# ILAT - STAR
print("\n===== ILAT STAR =====")
for i in range(5, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(i):
        print("*", end="")
    print()
```

```python
# ILAT - UPPER - ROW
print("\n===== ILAT UPPER ROW =====")
for i in range(5, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(i):
        print(chr(i + 64), end="")
    print()
```

```python
# ILAT - UPPER - COL
print("\n===== ILAT UPPER COL =====")
for i in range(5, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(1, i):
        print(chr(k + 64), end="")
    print()
```

```python
# ILAT - LOWER - ROW
print("\n===== ILAT LOWER ROW =====")
for i in range(5, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(i):
        print(chr(i + 96), end="")
    print()
```

```python
# ILAT - LOWER - COL
print("\n===== ILAT LOWER COL =====")
for i in range(5, 0, -1):
    for j in range(6, i, -1):
        print(" ", end="")
    for k in range(1, i):
        print(chr(k + 96), end="")
    print()
```

```python
# ILAT - NAME - ROW
print("\n===== ILAT NAME ROW =====")
name = "bharath"
for i in range(len(name) - 1, -1, -1):
    for j in range(len(name) - 1, i, -1):
        print(" ", end="")
    for k in range(i + 1):
        print(name[i], end="")
    print()
```

```python
# ILAT - NAME - COL
print("\n===== ILAT NAME COL =====")
name = "bharath"
for i in range(len(name) - 1, -1, -1):
    for j in range(len(name) - 1, i, -1):
        print(" ", end="")
    for k in range(i + 1):
        print(name[k], end="")
    print()
```
