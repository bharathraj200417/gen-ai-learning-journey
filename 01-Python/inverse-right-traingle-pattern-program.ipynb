```python
# IRAT - NUMBER - ROW
print("===== IRAT NUMBER ROW =====")
for i in range(6, 0, -1):
    for j in range(i):
        print(i, end="")
    print()
```

```python
# IRAT - NUMBER - COL
print("\n===== IRAT NUMBER COL =====")
for i in range(6, 0, -1):
    for j in range(i):
        print(j, end="")
    print()
```

```python
# IRAT - STAR
print("\n===== IRAT STAR =====")
for i in range(6, 0, -1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
# IRAT - UPPER - ROW
print("\n===== IRAT UPPER ROW =====")
for i in range(6, 0, -1):
    for j in range(i):
        print(chr(i + 64), end="")
    print()
```

```python
# IRAT - UPPER - COL
print("\n===== IRAT UPPER COL =====")
for i in range(6, 0, -1):
    for j in range(1, i):
        print(chr(j + 64), end="")
    print()
```

```python
# IRAT - LOWER - ROW
print("\n===== IRAT LOWER ROW =====")
for i in range(6, 0, -1):
    for j in range(i):
        print(chr(i + 96), end="")
    print()
```

```python
# IRAT - LOWER - COL
print("\n===== IRAT LOWER COL =====")
for i in range(6, 0, -1):
    for j in range(1, i):
        print(chr(j + 96), end="")
    print()
```

```python
# IRAT - NAME - ROW
print("\n===== IRAT NAME ROW =====")
name = "bharath"
for i in range(len(name) - 1, -1, -1):
    for j in range(i + 1):
        print(name[i], end="")
    print()
```

```python
# IRAT - NAME - COL
print("\n===== IRAT NAME COL =====")
name = "bharath"
for i in range(len(name), 0, -1):
    for j in range(i):
        print(name[j], end="")
    print()
```
