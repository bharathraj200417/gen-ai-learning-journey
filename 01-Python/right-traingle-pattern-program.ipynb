```python
# RAT - NUMBER - ROW
print("===== RAT NUMBER ROW =====")
for i in range(1, 6):
    for j in range(i):
        print(i, end="")
    print()
```

```python
# RAT - NUMBER - COL
print("\n===== RAT NUMBER COL =====")
for i in range(1, 6):
    for j in range(i):
        print(j, end="")
    print()
```

```python
# RAT - STAR
print("\n===== RAT STAR =====")
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
```

```python
# RAT - UPPER - ROW
print("\n===== RAT UPPER ROW =====")
for i in range(1, 6):
    for j in range(i):
        print(chr(i + 64), end="")
    print()
```

```python
# RAT - UPPER - COL
print("\n===== RAT UPPER COL =====")
for i in range(1, 7):
    for j in range(1, i):
        print(chr(j + 64), end="")
    print()
```

```python
# RAT - LOWER - ROW
print("\n===== RAT LOWER ROW =====")
for i in range(1, 6):
    for j in range(i):
        print(chr(i + 96), end="")
    print()
```

```python
# RAT - LOWER - COL
print("\n===== RAT LOWER COL =====")
for i in range(1, 7):
    for j in range(1, i):
        print(chr(j + 96), end="")
    print()
```

```python
# RAT - NAME - ROW
print("\n===== RAT NAME ROW =====")
name = "bharath"
for i in range(len(name)):
    for j in range(i + 1):
        print(name[i], end="")
    print()
```

```python
# RAT - NAME - COL
print("\n===== RAT NAME COL =====")
name = "bharath"
for i in range(len(name)):
    for j in range(i + 1):
        print(name[j], end="")
    print()
```
