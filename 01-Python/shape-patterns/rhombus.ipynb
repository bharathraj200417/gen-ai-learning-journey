```python
def rhombus():
    rhombus=((1,3),(1,4),(1,5),(1,6),
             (3,2),(3,5),(5,1),
            (5,2),(5,3),(5,4))
    print("=====RHOMBBUS=====")
    for i in range(1,8):
        for j in range(1,8):
            if(i,j) in rhombus:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
rhombus()
```
