```python
def pentagon():
    pentagon=((1,3),(2,1),(2,5),(4,2),
              (4,3),(4,4))
    print("=====PENTAGON=====")
    for i in range(1,10):
        for j in range(1,10):
            if(i,j) in pentagon:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
pentagon()
```
