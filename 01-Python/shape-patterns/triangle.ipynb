```python
def triangle():
    triangle=((1,4),(2,3),(2,5),(3,2),(3,6),
            (4,1),(4,2),(4,3),(4,4),(4,5),
            (4,5),(4,6),(4,7))
    print("=====TRIANGLE=====")
    for i in range(1,8):
        for j in range(1,8):
            if(i,j) in triangle:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
triangle()
```
