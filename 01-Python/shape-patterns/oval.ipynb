```python
def oval():
    oval=((1,5),(2,2),(2,8),(3,1),
             (3,9),(4,2),(4,8),(5,5))
    print("=====OVAL=====")
    for i in range(1,10):
        for j in range(1,10):
            if(i,j) in oval:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
oval()
```
