```python
def decagon():
    decagon=((1,4),(1,6),(2,2),(2,8),(4,1),
             (4,9),(6,2),(6,8),(8,4),
             (8,6))
    print("=====DECAGON=====")
    for i in range(1,10):
        for j in range(1,10):
            if(i,j) in decagon:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
decagon()
```
