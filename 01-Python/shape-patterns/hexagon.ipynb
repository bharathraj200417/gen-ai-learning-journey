```python
def hexagon():
    hexagon=((1,3),(1,4),(1,5),(2,2),
             (2,6),(3,1),(3,7),(4,2),
             (4,6),(5,3),(5,4),(5,5))
    print("=====HEXAGON=====")
    for i in range(1,10):
        for j in range(1,10):
            if(i,j) in hexagon:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
hexagon()
```
