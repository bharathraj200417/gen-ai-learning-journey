```python
def octagon():
    octagon=((1,3),(1,4),(1,5),(2,2),
             (2,6),(3,1),(3,7),(4,1),
             (4,7),(5,1),(5,7),(6,2),
             (6,6),(7,3),(7,4),(7,5))
    print("=====OCTAGON=====")
    for i in range(1,10):
        for j in range(1,10):
            if(i,j) in octagon:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
octagon()
```
