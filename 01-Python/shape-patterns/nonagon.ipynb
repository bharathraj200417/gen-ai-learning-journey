```python
def nonagon():
    nonagon=((1,5),(2,3),(2,7),(4,1),
             (4,9),(6,2),(6,8),(8,4),
             (8,6))
    print("=====NONAGON=====")
    for i in range(1,10):
        for j in range(1,10):
            if(i,j) in nonagon:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
nonagon()
```
