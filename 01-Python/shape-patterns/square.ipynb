```python
def square():
    square=((1,1),(1,2),(1,3),(1,4),
            (2,1),(2,4),(3,1),(3,4),
            (4,1),(4,2),(4,3),(4,4))
    print("=====SQUARE=====")
    for i in range(1,5):
        for j in range(1,5):
            if(i,j) in square:
                print("*",end=' ')
            else:
                print(' ',end=' ')
        print()
square()
```
