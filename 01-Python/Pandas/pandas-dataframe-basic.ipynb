```python
import pandas as pd
import numpy as np
print("libraries imported sucessfully")
```

```python
#creating series from a list with Intrinstic index
company_year = pd.Series(['IBM', 'Microsoft', 'Apple Inc.', 'Amazon.com', 'Alphabet Inc.'])
print("series with Intrinstic Index:")
print(company_year)
```

```python
#creating series from a list with Extrinstic index
company_year = pd.Series([1911,1975,1976,1994,1998],
            index=['IBM', 'Microsoft', 'Apple Inc.', 'Amazon.com', 'Alphabet Inc.']) 
print("series with Extrinstic Index:")
print(company_year)
```

```python
#creating DataFrame from a dictionary
data = {'Company':['Microsoft','Alpahabet Inc.','Apple Inc.'],
        'Founded':[1975,1998,1976],
        'Founder':['Bill Gates', 'Larry Page', 'Steve Jobs']}
df = pd.DataFrame(data)
print("DataFrame:\n",df)
```

```python
#usefull methods to explore data
print("First Two rows:\n", df.head(2))
print("Last Two rows:\n", df.tail(2))
print("Information:")
print(df.info())
#print("Describe:", df.describe())
```

```python
#Loc[] (access by label)  
print("using Loc[]:\n", df.loc[1])
print("using Loc[] specified cell:\n", df.loc[0, 'Founder'])
#iloc[] (acess by position)
print("using iloc[]:\n" , df.iloc[0:2])
```
